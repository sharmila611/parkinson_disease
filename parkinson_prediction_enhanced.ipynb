```python
# Remove the name column
df = df.drop("name", axis=1)

df.head()
```

```python
# Features (Input)
X = df.drop("status", axis=1)

# Target (Output)
y = df["status"]
```

```python
from sklearn.model_selection import train_test_split

# Stratified split preserves the Healthy/Parkinson class ratio in train/test.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape)
print("Testing samples :", X_test.shape)
print("\nTraining class distribution:")
print(y_train.value_counts(normalize=True))
print("\nTesting class distribution:")
print(y_test.value_counts(normalize=True))
```

```python
print("Training samples:", X_train.shape)
print("Testing samples :", X_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

# Balanced class weights reduce the effect of the 0/1 class imbalance.
rf = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)

rf.fit(X_train, y_train)
```

```python
y_pred = rf.predict(X_test)
```

## Enhanced novelty layer — voice explainability and imbalance handling


```python
# ============================================================
# ENHANCED NOVELTY LAYER — VOICE MODEL
# ============================================================
# Adds:
# 1) stratified split + class-balanced RF
# 2) F1 / ROC-AUC / PR-AUC
# 3) SHAP explainability
# 4) permutation/feature importance
# 5) calibrated probability export for multimodal fusion
#
# Run this section AFTER the cells above.
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, average_precision_score, classification_report,
    confusion_matrix
)
from sklearn.inspection import permutation_importance

# ---- Refit the final balanced voice RF ----
voice_rf_balanced = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)
voice_rf_balanced.fit(X_train, y_train)

voice_test_prob = voice_rf_balanced.predict_proba(X_test)[:, 1]
voice_test_pred = (voice_test_prob >= 0.5).astype(int)

voice_metrics = {
    "Accuracy": accuracy_score(y_test, voice_test_pred),
    "Precision": precision_score(y_test, voice_test_pred, zero_division=0),
    "Recall": recall_score(y_test, voice_test_pred, zero_division=0),
    "F1": f1_score(y_test, voice_test_pred, zero_division=0),
    "ROC-AUC": roc_auc_score(y_test, voice_test_prob),
    "PR-AUC": average_precision_score(y_test, voice_test_prob),
}
print("Balanced Voice RF metrics:")
display(pd.DataFrame([voice_metrics]))

print("\nClassification report:")
print(classification_report(
    y_test, voice_test_pred,
    target_names=["Healthy", "Parkinson's"],
    zero_division=0
))

# ---- SHAP: global explanation of the voice RF ----
# Install once if needed: pip install shap
try:
    import shap

    # TreeExplainer is appropriate for Random Forest.
    explainer = shap.TreeExplainer(voice_rf_balanced)
    shap_values = explainer.shap_values(X_test)

    # SHAP versions differ for binary tree models.
    if isinstance(shap_values, list):
        shap_for_pd = shap_values[1]
    else:
        shap_for_pd = shap_values
        if shap_for_pd.ndim == 3:
            shap_for_pd = shap_for_pd[:, :, 1]

    plt.figure()
    shap.summary_plot(
        shap_for_pd,
        X_test,
        plot_type="bar",
        show=False
    )
    plt.title("Voice RF — Global SHAP Feature Importance")
    plt.tight_layout()
    plt.show()

    plt.figure()
    shap.summary_plot(
        shap_for_pd,
        X_test,
        show=False
    )
    plt.title("Voice RF — SHAP Impact on Parkinson's Prediction")
    plt.tight_layout()
    plt.show()

except ImportError:
    print("SHAP is not installed. Run: pip install shap")
except Exception as e:
    print("SHAP could not be generated:", repr(e))

# ---- Model-agnostic permutation importance ----
perm = permutation_importance(
    voice_rf_balanced,
    X_test,
    y_test,
    n_repeats=10,
    random_state=42,
    scoring="f1"
)

perm_df = pd.DataFrame({
    "Feature": X_test.columns,
    "Importance_Mean": perm.importances_mean,
    "Importance_STD": perm.importances_std
}).sort_values("Importance_Mean", ascending=False)

display(perm_df.head(15))

plt.figure(figsize=(9, 6))
top = perm_df.head(12).sort_values("Importance_Mean")
plt.barh(top["Feature"], top["Importance_Mean"])
plt.xlabel("Mean decrease in F1 after permutation")
plt.title("Voice RF — Permutation Feature Importance")
plt.tight_layout()
plt.show()

# ---- ROC curve ----
from sklearn.metrics import RocCurveDisplay, PrecisionRecallDisplay

RocCurveDisplay.from_predictions(y_test, voice_test_prob)
plt.title(f"Voice RF ROC Curve (AUC={voice_metrics['ROC-AUC']:.3f})")
plt.tight_layout()
plt.show()

PrecisionRecallDisplay.from_predictions(y_test, voice_test_prob)
plt.title(f"Voice RF Precision-Recall Curve (AP={voice_metrics['PR-AUC']:.3f})")
plt.tight_layout()
plt.show()

# ---- Save the corrected model used by the multimodal notebook ----
import joblib
joblib.dump(voice_rf_balanced, "parkinsons_model_balanced.pkl")
print("Saved: parkinsons_model_balanced.pkl")

# Optional: save test probabilities for a future *patient-aligned* fusion dataset.
voice_probability_export = pd.DataFrame({
    "voice_probability": voice_test_prob,
    "label": np.asarray(y_test)
})
voice_probability_export.to_csv("voice_test_probabilities.csv", index=False)
print("Saved: voice_test_probabilities.csv")
```
